```python
import molsysmt as msm
import topomt as tmt
import molsysviewer_topomt as msv_tmt

molecular_system = msm.convert(tmt.demo['synthetic']['tube_channel.pdb'])
topography = tmt.get_topography(molecular_system, method='dfnd', probe_radius=1.4)

view = msv_tmt.new_view(topography)
view
```

```python
msv_tmt.show_dfnd_components(
      view,
      topography,
      representation='tetrahedra',
      component_ids=['WET-1'],
      draw_faces=True,
      draw_edges=True,
      alpha=0.25,
  )
```

```python
 msv_tmt.show_dfnd_components(
      view,
      topography,
      representation='spheres',
      component_ids=['WET-2'],
      alpha=0.35,
  )
```
