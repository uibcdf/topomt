# Smoke test

```python
import molsysmt as msm
import topomt as tmt
import molsysviewer_topomt as msv_tmt
```

```python
def make_view(system_key, probe_radius=1.4):
    molecular_system = msm.convert(tmt.demo['synthetic'][system_key])
    topography = tmt.get_topography(
        molecular_system,
        method='dfnd',
        probe_radius=probe_radius,
    )
    view = msv_tmt.new_view(topography)
    return molecular_system, topography, view
```

```python
molecular_system, topography, view = make_view('tube_channel.pdb')
```
