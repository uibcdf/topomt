## Receive public objects and physical patches separately

The table executes `get_topography(method='dfnd')` on the original inputs.
Native exterior links and public `Mouth` children retain their compatibility
semantics. Physical strict-clearance hull patches have separate clipped supports
and path certificates: their counts must not be substituted for one another.

```python
probes = [1.4, 3.0, 5.0]
expected_patches = [1, 0, 0]
topographies = {probe: study.receive(case_directory, probe) for probe in probes}
observed = {probe: study.inspect(topo) for probe, topo in topographies.items()}
rows = []
for probe, expected in zip(probes, expected_patches, strict=True):
    topo = topographies[probe]
    audit = observed[probe]
    boundary = audit['physical_boundary']
    assert boundary['status'] == 'resolved' and boundary['n_patches'] == expected
    rows.append([probe, ', '.join(type(f).__name__ for f in topo.values()) or 'Empty registry',
                 len(topo.dfnd.raw['external_links']),
                 len(topo.get_features(by='type', value='mouth')),
                 boundary['n_patches'], boundary['status']])
table(['Probe (Å)', 'Public objects', 'Native links', 'Mouth children',
       'Physical patches', 'Physical status'], rows)
topography = topographies[1.4]
audit = observed[1.4]
```

## Recover support and context

The audit verifies a complete, disjoint native wet/dry partition, mapped source
quadruples, reciprocal coast incidence/triangle areas, rooted graph distances
and public parent/child/catalog provenance. Lining atom lists cannot replace
carrier support. Graph depth without an exterior root uses an explicit zero
convention, not a measurement of physical burial.

```python
table(['Native component', 'Resident cells', 'Connectors', 'Depth context'],
      [[c['component_id'], len(c['resident_node_indices']),
        len(c['connector_node_indices']), c['depth_reference']]
       for c in audit['wet_components']])
print('Complete node inventory:', audit['node_inventory'])
print('Dry banks / coast faces:', len(audit['dry_components']), len(audit['coast_faces']))
component = topography.dfnd.dfn.components.wet[0]
print('First recoverable source tetrahedra:', component.tetrahedron_support[:3])
print('Shape/signature/catalog:', [(f.shape_type, f.signature, f.classification['name'])
      for f in topography.values() if f.feature_type != 'mouth'])
if component.transit_connector_node_indices:
    connector = component.transit_connector_node_indices[0]
    node = topography.dfnd.mesh.neighbors(connector)[0]
    graph = topography.dfnd.dfn.graph
    assert connector in graph.neighbors(node, side='dry', component_side='wet')
    print('Nonresident wet connector selected explicitly:', connector)
print('Participant/interface validation:', audit['public_molecular_interface_validation'])
print('Versioned public interchange:', audit['versioned_public_topography_interchange'])
```

## Replay independent geometric references

These checkers verify retained certificates using the frozen coordinates and
radii. They do not use DFND outputs as their oracle or run discovery optimizers.
The closed shell includes the independent 13-patch small-probe reference; both
shells include their 0.005 Å origin-access/enclosure bracket. These are not exact

```python
verified = study.references(case_directory)
for name, result in verified.items():
    if name == 'access_bracket':
        print(name, {key: result[key] for key in
              ('origin_access_lower_probe', 'origin_enclosed_upper_probe', 'bracket_width')})
    else:
        print(name, {key: result[key] for key in
              ('conclusion', 'opening_component_count', 'origin_path_status',
               'origin_connected_patch_count') if key in result})
```

## Reporting and ownership

Suppressing a reported object cannot close a physical opening. A shallow copy
has its own registry while retaining the documented shared analysis payload;
existing deep-copy, input-mutation and source-bound native restoration guards
cover their separate contracts. No public saved-result schema is invented here.

```python
cutoff = len(component.node_indices) + 1
suppressed = study.receive(case_directory, 1.4, min_size=cutoff)
assert not suppressed
assert suppressed.dfnd.raw['parameters']['result_key'] == topography.dfnd.raw['parameters']['result_key']
assert suppressed.dfnd.raw['boundary_connectivity'] == topography.dfnd.raw['boundary_connectivity']
copied = topography.copy(deep=False)
assert copied is not topography
assert all(feature.topography is copied for feature in copied.values())
print('Reporting suppression preserves raw/physical identity; copy registry is independently bound.')
```
